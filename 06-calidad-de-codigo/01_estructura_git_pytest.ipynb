## Fase 2: Logica y Documentacion

**Recordatorio:** un docstring explica que hace una funcion, que recibe y que devuelve, sin que quien lo lea tenga que revisar el codigo. Podes usar estilo **Google** o **NumPy**, cualquiera de los dos es valido siempre que cubra parametros, tipos de datos y proposito.

Estilo Google (referencia rapida):

```python
def funcion(param1, param2):
    """Resumen breve de que hace la funcion.

    Args:
        param1 (tipo): descripcion.
        param2 (tipo): descripcion.

    Returns:
        tipo: descripcion de lo que devuelve.
    """
```

Estilo NumPy (referencia rapida):

```python
def funcion(param1, param2):
    """
    Resumen breve de que hace la funcion.

    Parameters
    ----------
    param1 : tipo
        descripcion.
    param2 : tipo
        descripcion.

    Returns
    -------
    tipo
        descripcion de lo que devuelve.
    """
```

**Tu tarea:**

Desarrolla `src/procesamiento.py` con la funcion `calcular_metricas_ventas(dataframe)`. El dataset de pedidos tiene columnas `pedido_id`, `cliente`, `monto` y `canal`. La funcion debe devolver un diccionario con:

- `total_facturado`: la suma de `monto`.
- `monto_promedio_pedido`: el `total_facturado` dividido entre la cantidad de pedidos.
- `cantidad_pedidos`: la cantidad de filas del dataframe.
- `cantidad_clientes_unicos`: la cantidad de clientes distintos.

Si el dataframe esta vacio, las cuatro metricas deben devolverse en cero, no lanzar un error. No olvides el docstring.

````python
%%writefile src/procesamiento.py
"""Modulo de procesamiento de metricas de ventas para pedidos de tienda online."""


def calcular_metricas_ventas(dataframe):
    """Calcula metricas de ventas a partir de un DataFrame de pedidos.

    Args:
        dataframe (pandas.DataFrame): DataFrame con columnas 'pedido_id',
            'cliente', 'monto' y 'canal'. Cada fila representa un pedido.

    Returns:
        dict: Diccionario con las siguientes claves:
            - total_facturado (float): suma total de la columna 'monto'.
            - monto_promedio_pedido (float): total_facturado dividido entre
              la cantidad de pedidos.
            - cantidad_pedidos (int): cantidad de filas del dataframe.
            - cantidad_clientes_unicos (int): cantidad de clientes distintos
              en la columna 'cliente'.
        Si el dataframe esta vacio, las cuatro metricas se devuelven en 0.
    """
    if dataframe.empty:
        return {
            "total_facturado": 0,
            "monto_promedio_pedido": 0,
            "cantidad_pedidos": 0,
            "cantidad_clientes_unicos": 0,
        }

    total_facturado = dataframe["monto"].sum()
    cantidad_pedidos = len(dataframe)
    monto_promedio_pedido = total_facturado / cantidad_pedidos
    cantidad_clientes_unicos = dataframe["cliente"].nunique()

    return {
        "total_facturado": total_facturado,
        "monto_promedio_pedido": monto_promedio_pedido,
        "cantidad_pedidos": cantidad_pedidos,
        "cantidad_clientes_unicos": cantidad_clientes_unicos,
    }
````

## Fase 3: Pruebas Unitarias

**Recordatorio:** una prueba con pytest es una funcion cuyo nombre empieza con `test_`, y que usa `assert` para verificar un resultado. Como minimo se pide un test de calculo (con datos conocidos) y un test de vacio (dataframe sin filas).

**Tu tarea:**

Crea `tests/test_procesamiento.py` con al menos dos pruebas para `calcular_metricas_ventas`:

- **Test de calculo**: verifica que, con un dataset de pedidos conocido, las cuatro metricas sean las esperadas.
- **Test de vacio**: verifica que, con un dataframe sin filas, la funcion devuelva las cuatro metricas en cero.

````python
%%writefile tests/test_procesamiento.py
import pandas as pd
import pytest

from src.procesamiento import calcular_metricas_ventas


def test_calcular_metricas_ventas_con_datos():
    """Verifica el calculo de metricas con un dataset conocido de pedidos."""
    df = pd.DataFrame({
        "pedido_id": [1, 2, 3],
        "cliente": ["Ana", "Luis", "Ana"],
        "monto": [50.0, 30.0, 20.0],
        "canal": ["web", "app", "web"],
    })

    resultado = calcular_metricas_ventas(df)

    assert resultado["total_facturado"] == 100.0
    assert resultado["monto_promedio_pedido"] == pytest.approx(33.333, rel=1e-3)
    assert resultado["cantidad_pedidos"] == 3
    assert resultado["cantidad_clientes_unicos"] == 2


def test_calcular_metricas_ventas_vacio():
    """Verifica que un dataframe vacio devuelva las cuatro metricas en cero."""
    df = pd.DataFrame(columns=["pedido_id", "cliente", "monto", "canal"])

    resultado = calcular_metricas_ventas(df)

    assert resultado["total_facturado"] == 0
    assert resultado["monto_promedio_pedido"] == 0
    assert resultado["cantidad_pedidos"] == 0
    assert resultado["cantidad_clientes_unicos"] == 0
````

## Entrega en Moodle

Antes de comprimir el proyecto, corramos una verificacion automatica de que los puntos pedidos esten completos.

````python
import os
import subprocess
import sys

checklist = []

def check(condicion, mensaje):
    estado = "OK" if condicion else "FALTA"
    checklist.append((estado, mensaje))

check(os.path.isdir("data"), "Carpeta data/")
check(os.path.isdir("src"), "Carpeta src/")
check(os.path.isdir("tests"), "Carpeta tests/")
check(os.path.isdir(".git"), "Repositorio Git inicializado (.git)")

gitignore_ok = False
if os.path.exists(".gitignore"):
    contenido = open(".gitignore").read()
    gitignore_ok = "__pycache__" in contenido and ".csv" in contenido
check(gitignore_ok, ".gitignore excluye __pycache__/ y archivos .csv")

requirements_ok = False
if os.path.exists("requirements.txt"):
    contenido = open("requirements.txt").read().lower()
    requirements_ok = "pandas" in contenido and "pytest" in contenido
check(requirements_ok, "requirements.txt incluye pandas y pytest")

docstring_ok = False
if os.path.exists("src/procesamiento.py"):
    sys.path.insert(0, ".")
    import importlib
    import src.procesamiento as procesamiento
    importlib.reload(procesamiento)
    doc = procesamiento.calcular_metricas_ventas.__doc__
    docstring_ok = bool(doc and len(doc.strip()) > 20)
check(docstring_ok, "calcular_metricas_ventas tiene un docstring")

tests_ok = False
if os.path.exists("tests/test_procesamiento.py"):
    resultado = subprocess.run(["python", "-m", "pytest", "tests/", "-q"])
    tests_ok = resultado.returncode == 0
check(tests_ok, "Las pruebas de pytest pasan correctamente")

print("Resultado de la verificacion:\n")
for estado, mensaje in checklist:
    print(f"[{estado}] {mensaje}")
````
